### 1. Sorting Arrays
#### 1.1 Introduction to Array Sorting
**Sorting** adalah proses mengurutkan elemen array, misal dari nilai terkecil ke terbesar. Fungsi `np.sort()` menghasilkan **array baru** yang sudah terurut sehingga array asli tidak berubah.

#### 1.2 Sorting a One-Dimensional Array (1D)
Untuk array 1D, `np.sort()` mengurutkan semua elemen secara menaik. Gunakan `[::-1]` pada hasilnya jika ingin urutan menurun.

````python
scores = np.array([88, 72, 95, 80, 76])

print("Original:", scores)
print("Ascending:", np.sort(scores))
print("Descending:", np.sort(scores)[::-1])
print("Original remains:", scores)
````

`[::-1]` → Membalik urutan array, sehingga menjadi descending.

#### 1.3 Sorting a Two-Dimensional Array (2D)
Pada array 2D, `np.sort()` secara default mengurutkan nilai **di setiap baris** (`axis=-1`, sama dengan `axis=1` untuk array 2D). Fungsi ini tidak mengurutkan seluruh baris berdasarkan satu kolom.

````python
measurement_matrix = np.array([[8, 3, 6],
                               [9, 2, 5],
                               [4, 7, 1]])

print("Original matrix:")
print(measurement_matrix)
print("Sorted within each row:")
print(np.sort(measurement_matrix))
````

#### 1.4 Sorting by Axis
Untuk array 2D:

- **axis=0**: mengurutkan nilai dari atas ke bawah pada **setiap kolom**.
- **axis=1**: mengurutkan nilai dari kiri ke kanan pada **setiap baris**.
- **axis=None**: mengurutkan seluruh elemen dan menghasilkan array 1D.

````python
print("Sorted by columns (axis=0):")
print(np.sort(measurement_matrix, axis=0))

print("Sorted by rows (axis=1):")
print(np.sort(measurement_matrix, axis=1))

print("All elements sorted:", np.sort(measurement_matrix, axis=None))
````

#### 1.5 Sorting Indices with `np.argsort()`
`np.argsort()` menghasilkan **indeks urutan**, bukan nilai yang langsung diurutkan. Indeks tersebut berguna jika nilai dan nama mahasiswa harus tetap berpasangan.

````python
students = np.array(["Ayu", "Budi", "Citra", "Doni"])
final_scores = np.array([81, 94, 76, 88])
order = np.argsort(final_scores)

print("Sorted scores:", final_scores[order])
print("Corresponding students:", students[order])
````

### 2. Filtering Arrays
#### 2.1 Understanding Boolean Masks
**Filtering** digunakan untuk memilih data yang memenuhi syarat tertentu. **Boolean mask** adalah array berisi `True` dan `False` yang dihasilkan dari operasi perbandingan.

````python
exam_scores = np.array([65, 78, 84, 92, 71, 88])
mask = exam_scores >= 80

print("Scores:", exam_scores)
print("Mask >= 80:", mask)
````

#### 2.2 Boolean Indexing
Gunakan `array[mask]` untuk mengambil elemen pada posisi yang bernilai `True`. Hasilnya hanya berisi elemen yang memenuhi syarat.

````python
print("Scores >= 80:", exam_scores[exam_scores >= 80])
print("Scores < 80:", exam_scores[exam_scores < 80])
````

#### 2.3 Boolean Indexing in a Two-Dimensional Array
Untuk matriks, kondisi dapat digunakan untuk memilih **baris** berdasarkan salah satu kolom. Perhatikan bahwa hasil pilihan beberapa baris tetap berupa array 2D.

````python
student_marks = np.array([[75, 82, 80],
                          [90, 93, 88],
                          [68, 74, 79],
                          [85, 87, 91]])

selected_rows = student_marks[student_marks[:, 0] >= 80]
print("Students with first score >= 80:")
print(selected_rows)
print("Selected shape:", selected_rows.shape)
````

````python
student_marks[:, 0]
````

````python
# urutannya adalah slicing dari semua baris (:) kolom pertama (0) yang first score lebih dari 80
# lalu logika boleannya untuk menyaring memilih baris baris sesuai kondisi booelan dengan semua kolom terpilih [kondisi,]
student_marks[student_marks[:, 0] >= 80]
````

#### 2.4 Fancy Indexing
**Fancy indexing** memilih elemen menggunakan daftar indeks tertentu. Berbeda dengan Boolean indexing, kita menentukan **posisi** elemen yang ingin diambil.

````python
values = np.array([10, 20, 30, 40, 50])
print("Selected positions:", values[[0, 2, 4]])

print("Selected rows:")
print(student_marks[[0, 3], :])
````

#### 2.5 Filtering with `np.where()`
`np.where(kondisi)` menghasilkan **indeks posisi** yang memenuhi kondisi. Sementara itu, `np.where(kondisi, nilai_benar, nilai_salah)` memilih nilai berdasarkan kondisi, bukan hanya memfilter elemen.

````python
rainfall = np.array([0, 12, 8, 25, 4, 18])
print("Indices with rainfall >= 10:", np.where(rainfall >= 10)[0])
print("Values with rainfall >= 10:", rainfall[np.where(rainfall >= 10)])

print("Rain / No rain:", np.where(rainfall > 0, "Rain", "No rain"))
````

````python
np.where(rainfall >= 10)
````

#### 2.6 Conditional Selection with Multiple Conditions
Gunakan `&` untuk `AND`, `|` untuk `OR`, dan `~` untuk `NOT`. Setiap kondisi harus diberi tanda kurung. Jangan menggunakan `and` atau `or` untuk membandingkan seluruh array NumPy.

````python
temperatures = np.array([26, 29, 32, 35, 28, 31])

print("28 to 32:", temperatures[(temperatures >= 28) & (temperatures <= 32)])
print("Below 28 or above 32:", temperatures[(temperatures < 28) | (temperatures > 32)])
print("Not above 30:", temperatures[~(temperatures > 30)])
````

### 3. Updating Arrays
#### 3.1 Updating Individual Array Elements
Nilai array bisa diubah menggunakan indeks, misalnya `array[1] = 10`. Gunakan `.copy()` jika ingin mengubah array baru tanpa mengubah data aslinya.

````python
original_data = np.array([10, 20, 30, 40])
updated_data = original_data.copy()
updated_data[1] = 25

print("Original:", original_data)
print("Updated:", updated_data)
````

tanpa `.copy` maka nanti saat mengupdate data nilai original data juga akan berubah

#### 3.2 Updating Rows and Columns
Pada array 2D, gunakan `array[baris, kolom]` untuk satu nilai, `array[baris, :]` untuk satu baris, dan `array[:, kolom]` untuk satu kolom. Jumlah nilai pengganti harus sesuai dengan posisi yang dipilih, kecuali jika menggunakan satu nilai skalar.

````python
marks = np.array([[70, 75, 80],
                  [82, 86, 88],
                  [90, 91, 93]])

marks[0, 0] = 78
marks[1, :] = [84, 87, 90]
marks[:, 2] = 95

print("Updated marks:")
print(marks)
````

#### 3.3 Conditional Replacement
Nilai yang memenuhi kondisi tertentu dapat diganti sekaligus dengan **Boolean indexing** atau `np.where()`. Untuk menghindari perubahan data asli, salin array terlebih dahulu.

````python
measurements = np.array([12, -2, 18, -5, 9])
cleaned = measurements.copy()
cleaned[cleaned < 0] = 0

print("Original:", measurements)
print("Updated with mask:", cleaned)
print("Updated with np.where:", np.where(measurements < 0, 0, measurements))
````

`np.where(measurements < 0, 0, measurements)` artinya jika <0 diberi nilai 0 jika tidak dikembalikan ke nilai aslinya

#### 3.4 Adding Rows with np.concatenate()
`np.concatenate((a, b), axis=0)` menggabungkan **baris**. Untuk matriks, jumlah **kolom** kedua array harus sama. Fungsi ini menghasilkan array baru.

````python
sales = np.array([[12, 15],
                  [10, 18]])
new_row = np.array([[14, 16]])

sales_with_row = np.concatenate((sales, new_row), axis=0)
print("Added row:")
print(sales_with_row)
print("New shape:", sales_with_row.shape)
````

#### 3.5 Adding Columns with `np.concatenate()`
Gunakan `axis=1` untuk menggabungkan **kolom**. Jumlah **baris** harus sama. Kolom baru perlu berbentuk 2D, misalnya `(3, 1)`, bukan `(3,)`.

````python
new_column = np.array([[20], [21], [22]])
sales_with_column = np.concatenate((sales_with_row, new_column), axis=1)

print("Added column:")
print(sales_with_column)
print("New shape:", sales_with_column.shape)
````

#### 3.6 Deleting Rows and Columns with np.delete()
`np.delete(array, indeks, axis=0)` menghapus **baris**, sedangkan `axis=1` menghapus **kolom**. Hasilnya adalah array baru; array asli tetap tersedia.

````python
print("Delete second row:")
print(np.delete(sales_with_column, 1, axis=0))
print("Delete third column:")
print(np.delete(sales_with_column, 2, axis=1))
````

#### 3.7 Shape and Dimension Compatibility
Saat menggabungkan array, dimensi (`ndim`) harus sama. Untuk `axis=0`, jumlah kolom harus sama; untuk `axis=1`, jumlah baris harus sama. `reshape()` membantu mengubah vektor 1D menjadi baris atau kolom 2D.

````python
matrix = np.arange(1, 7).reshape(2, 3)
row_values = np.array([7, 8, 9]).reshape(1, 3)
column_values = np.array([10, 11]).reshape(2, 1)

print("Matrix shape:", matrix.shape)
print("New row shape:", row_values.shape)
print("New column shape:", column_values.shape)

print("With row:")
print(np.concatenate((matrix, row_values), axis=0))

print("With column:")
print(np.concatenate((matrix, column_values), axis=1))
````

**Common mistakes**: Menambahkan kolom berbentuk `(2,)` ke matriks `(2, 3)` dengan `axis=1` tidak dapat dilakukan langsung. Ubah dahulu menjadi `(2, 1)`. Jangan menggunakan `np.delete()` tanpa axis jika ingin mempertahankan bentuk matriks karena hasil default-nya diratakan menjadi 1D.

### 4. Array Aggregation
#### 4.1 Sum, Minimum, Maximum, and Mean
**Aggregation** merangkum banyak nilai menjadi satu atau beberapa statistik. Fungsi dasar yang sering dipakai adalah `np.sum()`, `np.min()`, `np.max()`, dan `np.mean()`.

````python
daily_visitors = np.array([120, 135, 128, 142, 150])

print("Total:", np.sum(daily_visitors))
print("Minimum:", np.min(daily_visitors))
print("Maximum:", np.max(daily_visitors))
print("Mean:", np.mean(daily_visitors))
````

#### 4.2 Aggregation by Axis
Pada matriks, **axis=0** meringkas nilai dari berbagai baris untuk **setiap kolom**; **`axis=1`** meringkas nilai dari berbagai kolom untuk setiap baris.

**axis=0** totalan/rataan/... semua baris hasilnya per masing2 kolom ... per variabel

**axis=1** totalan/rataan/... semua kolom hasilnya per masing2 baris ... per observasi

````python
test_scores = np.array([[80, 85, 90],
                        [70, 75, 80],
                        [90, 92, 94]])

print("Total per test (axis=0):", np.sum(test_scores, axis=0))
print("Mean per student (axis=1):", np.mean(test_scores, axis=1))
print("Maximum per test (axis=0):", np.max(test_scores, axis=0))
````

#### 4.3 Cumulative Sum
`np.cumsum()` menghitung jumlah **kumulatif** secara bertahap. Nilai di setiap posisi adalah jumlah seluruh nilai sampai posisi tersebut. `axis` juga dapat digunakan untuk data 2D.

````python
weekly_sales = np.array([10, 15, 12, 18])

print("Weekly sales:", weekly_sales)
print("Cumulative sales:", np.cumsum(weekly_sales))

print("Cumulative scores across columns:")
print(np.cumsum(test_scores, axis=1))
````

#### 4.4 The `keepdims` Argument
Secara default, agregasi menghilangkan sumbu yang diringkas. `keepdims=True` mempertahankan sumbu tersebut dengan ukuran **1**, sehingga hasil lebih mudah dipakai untuk operasi lanjutan yang membutuhkan bentuk kompatibel.

````python
row_means = np.mean(test_scores, axis=1)
row_means_2d = np.mean(test_scores, axis=1, keepdims=True)

print("Without keepdims:", row_means, "shape:", row_means.shape)
print("With keepdims:")
print(row_means_2d)
print("Shape:", row_means_2d.shape)
````

#### 4.5 Applying Aggregation to Statistical Data
Misalkan setiap baris adalah satu hari dan setiap kolom adalah satu lokasi pengamatan curah hujan. Agregasi dapat menghitung total per lokasi dan rata-rata per hari.

````python
rainfall_matrix = np.array([[12, 8, 5],
                            [10, 15, 7],
                            [20, 11, 9],
                            [8, 6, 4]])

print("Total per location:", np.sum(rainfall_matrix, axis=0))
print("Mean per day:", np.mean(rainfall_matrix, axis=1))
print("Overall maximum:", np.max(rainfall_matrix))
````

### 5. Vectorized Operations
#### 5.1 Vectorization vs. Python Loops
**Vectorization** memungkinkan perhitungan langsung pada seluruh array tanpa menulis perulangan untuk setiap elemen. Kode umumnya lebih singkat dan sering lebih efisien untuk data numerik.

````python
values = np.array([2, 4, 6, 8])

# Python loop
loop_result = []
for value in values:
    loop_result.append(value * 2)

# NumPy vectorization
vector_result = values * 2

print("Loop:", loop_result)
print("Vectorized:", vector_result)
````

#### 5.2 Scalar Arithmetic Operations
Satu bilangan (**skalar**) dapat dijumlahkan, dikurangkan, dikalikan, atau dibagikan ke setiap elemen array. Operasi ini bekerja untuk array 1D maupun 2D.

````python
data = np.array([10, 20, 30])

print("Add 5:", data + 5)
print("Subtract 2:", data - 2)
print("Multiply by 3:", data * 3)
print("Divide by 2:", data / 2)
print("Squared:", data ** 2)
````

#### 5.3 Element-wise Array Mathematics
**Element-wise** berarti operasi dilakukan pada elemen yang **posisinya bersesuaian**. Contohnya, `a + b` menjumlahkan elemen pertama dengan pertama, kedua dengan kedua, dan seterusnya. Kedua array pada contoh berikut memiliki bentuk sama.

````python
a = np.array([10, 20, 30])
b = np.array([2, 4, 5])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Powers:", a ** 2)
````

#### 5.4 Element-wise Comparison Operations
Operator `>`, `<`, `>=`, `<=`, `==`, dan `!=` membandingkan setiap elemen dan menghasilkan array Boolean. Hasil ini dapat dipakai kembali untuk filtering atau pembaruan nilai.

````python
actual = np.array([75, 90, 82, 60])
target = np.array([80, 85, 80, 70])

print("Actual >= target:", actual >= target)
print("Actual == target:", actual == target)
print("Passed target:", actual[actual >= target])
````

#### 5.5 Shape Compatibility and Broadcasting
Operasi element-wise memerlukan bentuk yang **kompatibel**. **Broadcasting** memungkinkan array dengan bentuk berbeda dipasangkan apabila ukurannya sama atau salah satunya 1 saat dibandingkan dari dimensi paling kanan. Contohnya, vektor tiga nilai dapat ditambahkan pada setiap baris matriks dengan tiga kolom.

````python
matrix = np.array([[10, 20, 30],
                   [40, 50, 60]])
adjustment = np.array([1, 2, 3])

print("Matrix shape:", matrix.shape)
print("Adjustment shape:", adjustment.shape)
print("Element-wise result:")
print(matrix + adjustment)
````

**Important distinction**: Operator * pada dua array adalah perkalian **element-wise**, bukan perkalian matriks. Untuk perkalian matriks gunakan `@` atau `np.matmul()`. Operasi dengan bentuk yang tidak kompatibel akan menghasilkan `ValueError`.

#### 5.6 Applied Example: Standardizing Daily Measurements
Contoh sederhana berikut menghitung selisih setiap pengamatan dari rata-ratanya. Karena menggunakan operasi vektorisasi, kita tidak perlu menulis perulangan.

````python
observations = np.array([12.0, 15.0, 13.0, 20.0])
mean_value = np.mean(observations)
deviations = observations - mean_value

print("Mean:", mean_value)
print("Deviations:", deviations)
print("Sum of deviations:", np.sum(deviations))
````

4. Hitung rata-rata setiap baris menggunakan `keepdims=True`, lalu periksa bentuknya.

````python
visitors = np.array([[20, 25, 30],
                     [18, 22, 28],
                     [24, 27, 32]])

# 1
total = np.sum(visitors)
rata2 = np.mean(visitors)
print(f'1. Total dan rata-rata pengamatan adalah {total} dan {rata2}')

# 2
total = np.sum(visitors, axis=0)
rata2 = np.mean(visitors, axis=1)
print(f'2. Total setiap kolom: {total} \n rata-rata setiap baris: {rata2}')

# 3 
kum = np.cumsum(visitors, axis=0)
print('Jumlah kumulatif setiap baris:\n', kum)

# 4
rata2 = np.mean(visitors, axis=1, keepdims=True)
print(f'4. Rata-rata setiap baris menggunakan keepdims=True:\n {rata2} \n shape: {rata2.shape}')
````

### Essential Concepts

- **`np.sort()`** mengurutkan nilai dan **`np.argsort()`** menghasilkan indeks pengurutan.
- **`axis=0`** bekerja sepanjang baris untuk setiap kolom; **`axis=1`** bekerja sepanjang kolom untuk setiap baris.
- **Boolean masks**, **Fancy indexing**, dan **`np.where()`** membantu memilih data sesuai kondisi atau posisi.
- **Index assignment** dan **conditional replacement** digunakan untuk memperbarui nilai array.
- **`np.concatenate()`** menambahkan baris/kolom dan **`np.delete()`** menghapusnya dengan `axis` yang sesuai.
- **`.ndim`** dan **`.shape`** penting untuk memeriksa kompatibilitas bentuk saat menggabungkan array.
- **`np.sum()`**, **`np.min()`**, **`np.max()`**, dan **`np.mean()`** menghitung ringkasan data.
- **`np.cumsum()`** menghitung jumlah kumulatif; **`keepdims=True`** mempertahankan dimensi yang diringkas.
- **Vectorization** memungkinkan operasi skalar dan element-wise tanpa perulangan manual; perbandingan menghasilkan Boolean array.
