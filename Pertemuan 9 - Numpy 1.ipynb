## 1.3 Perbedaan List Python dan NumPy Arrays

**List Python** dapat menyimpan berbagai tipe data sekaligus.

**NumPy Array** hanya menyimpan elemen dengan tipe data yang sama, yang membuat lebih efisien untuk perhitungan numerik (angka).

**Misal** menggunakan operator `+` pada dua buah List, hasilnya adalah penggabungan (menyambung) kedua list tersebut. Namun, jika menggunakan operator `+` pada NumPy Array, hasilnya adalah penjumlahan matematika dari elemen-elemen yang posisinya bersesuaian.

```python
python_list = [2, 4, 6]
numpy_array = np.array([2, 4, 6])

print("Python list:", python_list)
print("NumPy array:", numpy_array)
print("Python type:", type(python_list))
print("NumPy type:", type(numpy_array))
```

```python
print(python_list + python_list)
print(numpy_array + numpy_array)
```

## 2. Membuat NumPy Arrays

### 2.1 Membuat Numpy Arrays dari Python Lists
Fungsi `np.array()` digunakan untuk mengubah data list Python menjadi NumPy Array. Jumlah dimensi ditentukan oleh struktur data yang dimasukkan.

- **Array 1D**: Berbentuk seperti garis atau vektor tunggal, cocok untuk data satu variabel (misal: nilai ujian mahasiswa)
- **Array 2D**: Berbentuk seperti matriks atau tabel dengan baris dan kolom.
- **Array 3D**: Kumpulan dari beberapa matriks 2D yang ditumpuk.
- **Array 4D**: Tingkatan lebih tinggi yang berisi sekumpulan array 3D, biasnya digunakan pada data ilmiah yang rumit.

Harus memiliki bentuk (`shape`) yang konsisten.
**Shape**: Ukuran NumPy Array yang menunjukkan jumlah elemen pada setiap dimensi.

### 2.2 Membuat Array Satu Dimensi (1D)
Array 1D untuk merepresentasikan pengukuran dari satu variabel, seperti nilai dari lima orang siswa.

```python
student_scores = [75, 80, 92, 86, 88]
scores_1d = np.array(student_scores)

print(scores_1d)
print("Dimensions:", scores_1d.ndim)
print("Shape:", scores_1d.shape)
```

### 2.3 Membuat Array Dua Dimensi (2D)
Array 2D merepresentasikan data dalam **baris dan kolom**. Pada contoh di materi, setiap baris mewakili seorang siswa, sedangkan setiap kolom mewakili penilaian yang berbeda.
Bentuk `(3, 3)` menunjukkan bahwa array tersebut memiliki tiga baris dan tiga kolom.

```python
student_assessments = [[80, 85, 90],
                       [75, 88, 84],
                       [92, 90, 95]]

scores_2d = np.array(student_assessments)

print(scores_2d)
print("Dimensions:", scores_2d.ndim)
print("Shape:", scores_2d.shape)
```

The shape `(3, 3)` indicates three rows and three columns.

### 2.4 Membuat Array Tiga Dimensi (3D)
Array 3D dapat dipahami sebagai **tumpukan matriks**.

```python
week_1 = np.array([[10, 12],
                   [14, 16]])
week_2 = np.array([[11, 13],
                   [15, 17]])

observations_3d = np.array([week_1, week_2])

print(observations_3d)
print("Dimensions:", observations_3d.ndim)
print("Shape:", observations_3d.shape)
```

Bentuk `(2, 2, 2)` menunjukkan ada dua matriks, yang masing-masing berisi **dua baris dan dua kolom**. Sumbu pertama mengidentifikasi matriks mana yang dipilih.

### 2.5 Membuat Array Empat Dimensi (4D)
Array 4D menambahkan tingkat pengelompokkan baru di atas struktur data yang sudah ada. Struktur seperti ini sering muncul pada data ilmiah dan kumpulan observasi multidimensi.
Dalam NumPy, setiap kali dimensi bertambah, data dibungkus ke dalam kelompok yang lebih besar.

```python
first_cube = np.array([[[1, 2], [3, 4]],
                       [[5, 6], [7, 8]]])
second_cube = np.array([[[11, 12], [13, 14]],
                        [[15, 16], [17, 18]]])

observations_4d = np.array([first_cube, second_cube])

print(observations_4d)
print("Dimensions:", observations_4d.ndim)
print("Shape:", observations_4d.shape)
```

### 2.6 Membuat Array dari Awal: `np.zeros()`
`np.zeros()` secara langsung membuat array yang seluruh isinya adalah angka nol.

```python
zero_vector = np.zeros(5)
zero_matrix = np.zeros((3, 4))

print("1D zeros:")
print(zero_vector)
print("2D zeros:")
print(zero_matrix)
```

### 2.7 Membuat Array Acak: `np.random.random()`
`np.random.random()` menghasilkan bilangan desimal acak antara **0 hingga kurang dari 1**. Penggunaan `seed` membuat hasil bilangan acak tetap sama setiap kali kode dijalankan.

```python
np.random.seed(42)
random_measurements = np.random.random((2, 3))

print(random_measurements)
print("Shape:", random_measurements.shape)
```

### 2.8 Membuat Deret (Sequences): `np.arange()`
`np.arange(start, stop, step)` digunakan untuk membuat deret angka dengan nilai awal `(start)`, batas akhir `(stop)`, dan jarak antarangka `(step)`. Nilai `stop` tidak disertakan. Secara default, `start = 0` dan `step = 1`.

```python
sequence_a = np.arange(5)
sequence_b = np.arange(2, 11, 2)
sequence_c = np.arange(-3, 4, 3)

print("Default sequence:", sequence_a)
print("Even-number sequence:", sequence_b)
print("Stepped sequence:", sequence_c)
```

## 3. Atribut Array dan Tipe Data

### 3.1 Dimensi Array dan Sumbu (Axes)
**Axis** adalah sumbu yang menunjukkan arah dimensi pada NumPy Array. Pada array 2D, `axis=0` menunjukkan arah baris dan `axis=1` menunjukkan arah kolom. Pada array 3D, terdapat tiga sumbu, yaitu lapisan, baris, dan kolom.

- `.ndim`: Jumlah dimensi atau sumbu.
- `.shape`: Ukuran setiap dimensi.
- `.size`: Jumlah seluruh elemen array.

| Attribute | Meaning | Example result |
| --------- | ------- | -------------- |
| `.ndim` | Jumlah dimensi atau sumbu | 2 |
| `.shape` | Ukuran setiap dimensi | (3, 4) |
| `.size` | Jumlah seluruh elemen array| 12 |
| `.dtype` | Tipe data umum dari elemen-elemennya | int64, float64, etc |

```python
measurement_matrix = np.array([[12, 15, 18, 21],
                               [14, 16, 20, 22],
                               [11, 13, 17, 19]])

print("Array:")
print(measurement_matrix)
print("ndim:", measurement_matrix.ndim)
print("shape:", measurement_matrix.shape)
print("size:", measurement_matrix.size)
print("dtype:", measurement_matrix.dtype)
```

### 3.2 Memahami Shape dan Size
Shape `(3, 4)` berarti array memiliki tiga baris dan empat kolom, sehingga ukurannya (*size*) adalah `3 × 4 = 12` elemen. Untuk array 3D atau 4D, total *size*-nya adalah hasil kali dari semua angka pada *shape*-nya.

```python
print("1D:", scores_1d.ndim, scores_1d.shape, scores_1d.size)
print("2D:", scores_2d.ndim, scores_2d.shape, scores_2d.size)
print("3D:", observations_3d.ndim, observations_3d.shape, observations_3d.size)
print("4D:", observations_4d.ndim, observations_4d.shape, observations_4d.size)
```

### 3.3 Tipe Data NumPy yang Umum `(dtype)` 
NumPy mencakup tipe data untuk bilangan bulat (`int`), bilangan desimal/pecahan (`float`), nilai Boolean (`bool`), dan teks (`str`). `.dtype` pada NumPy memberikan detail lebih lanjut, seperti jumlah bit yang digunakan. Karena NumPy array menggunakan tipe data yang seragam, menggabungkan jenis nilai yang berbeda dapat menyebabkan konversi tipe secara otomatis.

```python
integer_values = np.array([1, 2, 3])
float_values = np.array([1.5, 2.5, 3.5])
boolean_values = np.array([True, False, True])
string_values = np.array(["low", "medium", "high"])

print("Integer:", integer_values.dtype)
print("Float:", float_values.dtype)
print("Boolean:", boolean_values.dtype)
print("String:", string_values.dtype)
```

## 3.4 Menentukan Tipe Data Secara Eksplisit
Gunakan argumen dtype di dalam `np.array()` untuk meminta tipe data tertentu. Tipe numerik yang lebih kecil (seperti `float32` vs `float64`) akan memakan lebih sedikit memori, tetapi juga memiliki rentang atau presisi yang lebih terbatas

```python
measurements_float32 = np.array([1.25, 2.75, 3.50], dtype=np.float32)
counts_int32 = np.array([10, 20, 30], dtype=np.int32)

print(measurements_float32, measurements_float32.dtype)
print(counts_int32, counts_int32.dtype)
```

### 3.5 Konversi Tipe Data dengan `astype()`
Metode `.astype()` mengembalikan array yang sudah diubah ke tipe data yang ditentukan. Penerapan yang umum adalah mengubah teks (*string*) yang berisi angka menjadi nilai numerik sungguhan.

Hati-hati: mengubah bilangan desimal menjadi bilangan bulat (*integer*) akan membuang bagian pecahannya; ini tidak sama dengan membulatkan ke bilangan bulat terdekat.

```python
text_scores = np.array(["70", "85", "90"])
numeric_scores = text_scores.astype(np.int64)

print("Before:", text_scores, text_scores.dtype)
print("After:", numeric_scores, numeric_scores.dtype)
```

```python
continuous_values = np.array([1.9, 2.1, 3.8])
integer_conversion = continuous_values.astype(int)

print("Original:", continuous_values)
print("Converted:", integer_conversion)
```

### 3.6 Type Coercion
**Type coercion** terjadi ketika NumPy secara otomatis memilih satu tipe data umum untuk semua elemen yang digunakan saat membuat array. Contohnya termasuk mengubah bilangan bulat menjadi desimal jika ada satu angka desimal yang dimasukkan, atau mengubah nilai angka menjadi teks (*string*) jika dicampur dengan teks.

**Selalu periksa `.dtype` sebelum melakukan analisis numerik.**

```python
coercion_a = np.array([10, 20.5, 30])
coercion_b = np.array([True, False, 5])
coercion_c = np.array([1, "two", 3])

print(coercion_a, coercion_a.dtype)
print(coercion_b, coercion_b.dtype)
print(coercion_c, coercion_c.dtype)
```

## 4. Mengubah Bentuk Array (Reshaping)
### 4.1 Memahami bentuk Array (Shape)
*Shape* mendeskripsikan bagaimana elemen-elemen disusun. Penting diingat, array 1D dengan enam elemen memiliki shape `(6,)`, bukan `(1, 6)`. Array 2D berbentuk baris dan array 2D berbentuk kolom memiliki shape yang berbeda, meskipun keduanya berisi enam nilai yang sama.shaped 2D array have different shapes, even when they contain the same six values.

```python
base_values = np.arange(1, 7)

row_array = base_values.reshape(1, 6)
column_array = base_values.reshape(6, 1)

print("1D shape:", base_values.shape)
print("Row shape:", row_array.shape)
print(row_array)
print("Column shape:", column_array.shape)
print(column_array)
```

### 4.2 Mengubah Bentuk Array dengan `reshape()`
`.reshape()` mengubah susunan elemen tanpa mengubah total jumlah elemennya. Array 1D yang berisi 12 nilai dapat diubah menjadi `(3, 4)`, `(4, 3)`, atau `(2, 2, 3)` karena masing-masing bentuk tersebut memuat tepat 12 posisi. NumPy membaca elemen dalam urutan aslinya lalu menyusunnya ke dalam bentuk yang baru.

```python
values_12 = np.arange(1, 13)

matrix_3x4 = values_12.reshape(3, 4)
matrix_4x3 = values_12.reshape(4, 3)
array_2x2x3 = values_12.reshape(2, 2, 3)

print("Original:", values_12)
print("3 × 4:")
print(matrix_3x4)
print("4 × 3:")
print(matrix_4x3)
print("2 × 2 × 3:")
print(array_2x2x3)
```

### 4.3 Kompatibilitas Reshape
Total jumlah elemen **harus cocok** dengan bentuk yang diminta. Misalnya, array berukuran 12 tidak dapat diubah menjadi `(5, 3)`, karena bentuk tersebut membutuhkan 15 elemen. 

NumPy dapat menghitung satu dimensi yang hilang secara otomatis jika kamu memasukkan angka `-1`.

```python
reshaped_automatically = values_12.reshape(3, -1)

print(reshaped_automatically)
print("Inferred shape:", reshaped_automatically.shape)
```

### 4.4 Meratakan Array dengan `flatten()`
`.flatten()` mengubah array multidimensi menjadi **salinan array 1D yang baru**. Ini berguna saat analisismu membutuhkan seluruh elemen berada dalam satu deret lurus

```python
original_matrix = np.array([[1, 2],
                            [5, 7],
                            [6, 6]])
flattened_values = original_matrix.flatten()

print("Original:")
print(original_matrix)
print("Flattened:")
print(flattened_values)
print("Flattened shape:", flattened_values.shape)
```

### 4.5 `reshape()` vs `flatten()`
| Operasi | Tujuan | Jumlah dimensi |
| --------- | ------------ | ------------------------------------ |
| `reshape(new_shape)` | Menyusun ulang elemen ke dalam bentuk yang dipilih | shape yang baru `new_shape` |
| `flatten()` | Mengubah semua elemen menjadi array 1D baru | Selalu 1D |

## 5. Indeks dan Slicing Array Dasar
### 5.1 Mengindeks Array Satu Dimensi (1D)
**Indexing** memilih sebuah elemen berdasarkan posisinya. Python menggunakan **indeks berbasis nol**, jadi indeks `0` adalah elemen pertama, indeks `1` adalah elemen kedua, dan seterusnya.

Indeks negatif menghitung mundur dari akhir: `-1` adalah elemen terakhir.

```python
values = np.array([2, 4, 6, 8, 10])

print("First element:", values[0])
print("Fourth element:", values[3])
print("Last element:", values[-1])
```

### 5.2 Mengindeks Array Dua Dimensi (2D)
Array 2D membutuhkan **indeks baris** dan **indeks kolom** untuk memilih satu elemen. Sintaksnya adalah `array[baris, kolom]`. 

Kedua indeks dimulai dari nol.

```python
marks = np.array([[80, 85, 90],
                  [75, 88, 84],
                  [92, 90, 95]])

print(marks)
print("Row 1, column 2:", marks[1, 2])
print("Row 2, column 0:", marks[2, 0])
```

### 5.3 Memilih Baris dan Kolom
Gunakan indeks tunggal untuk memilih satu baris penuh `(array[baris])` atau gabungkan indeks dengan titik dua `:` untuk memilih satu kolom penuh `(array[:, kolom])`.

Tanda titik dua berarti **semua posisi** pada sumbu tersebut.

```python
print("First row:", marks[0])
print("Second row:", marks[1, :])
print("Third column:", marks[:, 2])
print("First column:", marks[:, 0])
```

### 5.4 Memotong (Slicing) Array Satu Dimensi (1D)
**Slicing** mengambil beberapa elemen yang berurutan. Sintaks umumnya adalah `array[start:stop]`.

- `start` diikutsertakan.
- `stop` tidak diikutsertakan (dikecualikan).
- Jika `start` dikosongkan, berarti dimulai dari awal array. 
- Jika `stop` dikosongkan, berarti mengambil sampai akhir array.

```python
sequence = np.array([10, 20, 30, 40, 50, 60])

print("Positions 1 to 3:", sequence[1:4])
print("First three:", sequence[:3])
print("From position 3 onward:", sequence[3:])
print("All elements:", sequence[:])
```

### 5.5 Slicing Array Dua Dimensi (2D)
Slice 2D menentukan rentang baris dan rentang kolom: `array[row_start:row_stop, col_start:col_stop]`. Ini berguna untuk mengekstrak bagian kecil (submatriks) dari sebuah matriks.

```python
number_grid = np.arange(1, 26).reshape(5, 5)

print("Original matrix:")
print(number_grid)
print("Rows 1–3, columns 2–4:")
print(number_grid[1:4, 2:5])
```

### 5.6 Slicing dengan Langkah (Steps)
Slice dapat menyertakan argumen ketiga: `array[start:stop:step]`. step menentukan jarak interval antar posisi yang dipilih.

- `::2` memilih setiap elemen kedua, dimulai dari awal.
- `1::2` memilih setiap elemen kedua, dimulai dari indeks 1.
- `[::-1]` membalikkan urutan array 1D.

Langkah (steps) ini juga dapat diterapkan secara terpisah pada sumbu baris dan kolom pada array 2D.

```python
print("Every second element:", sequence[::2])
print("Every second from index 1:", sequence[1::2])
print("Reversed:", sequence[::-1])
print("Every second row and column:")
print(number_grid[::2, ::2])
```

### 5.8 Kesalahan Umum dalam Indeks dan Slicing
1. **Off-by-one error (Kesalahan batas)**: `a[1:4]`memilih indeks 1, 2, dan 3, bukan indeks 4.
2. **Mixing row and column order (Tertukar urutan baris dan kolom)**: `a[2, 1]` berarti indeks baris 2 dan indeks kolom 1.
3. **Confusing element and row selection (Memilih elemen vs memilih baris)**: `a[1]` mengembalikan seluruh baris dari array 2D, sedangkan `a[1, 1]` mengembalikan satu elemen saja.
4. **Index out of range (Indeks di luar jangkauan)**: array berisi 5 elemen hanya memiliki indeks positif yang valid dari 0 hingga 4.
5. **Confusing 1D and 2D shapes (Memilih antara shape 1D dan 2D)**: a 1D array of size 5 has shape `(5,)`, not `(1, 5)`.

**Perbedaan Konseptual**: Indexing biasanya memilih satu posisi individu; sedangkan slicing memilih rentang posisi.
